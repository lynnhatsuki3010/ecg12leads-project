```python
import numpy as np
from glob import glob
import matplotlib.pyplot as plt

# Load nhiều samples
feat_files = glob(r"E:\NCKH - 2026\ECG Project\data\processed\hand_features_only_090125\*.npy")[:1000]
all_feats = [np.load(f) for f in feat_files]
all_feats = np.array(all_feats)  # (1000, 70)

print(f"Shape: {all_feats.shape}")
print(f"\nPer-feature statistics:")
print("="*80)

for i in range(70):
    feat_vals = all_feats[:, i]
    print(f"Feature {i:2d}: min={feat_vals.min():8.2f}, max={feat_vals.max():8.2f}, "
          f"mean={feat_vals.mean():8.2f}, std={feat_vals.std():8.2f}, "
          f"zero%={100*(feat_vals==0).sum()/len(feat_vals):5.1f}%")

# Tìm features có scale lớn
print("\n" + "="*80)
print("🚨 FEATURES WITH HUGE SCALE (max > 500):")
for i in range(70):
    if all_feats[:, i].max() > 500:
        print(f"  Feature {i}: max = {all_feats[:, i].max():.2f}")

print("\n🚨 FEATURES WITH NEAR-ZERO VARIANCE:")
for i in range(70):
    if all_feats[:, i].std() < 0.01:
        print(f"  Feature {i}: std = {all_feats[:, i].std():.4f} (almost constant!)")

print("\n🚨 FEATURES THAT ARE ALWAYS 0:")
for i in range(70):
    if (all_feats[:, i] == 0).sum() / len(all_feats) > 0.95:
        print(f"  Feature {i}: {100*(all_feats[:, i]==0).sum()/len(all_feats):.1f}% zeros")
```
