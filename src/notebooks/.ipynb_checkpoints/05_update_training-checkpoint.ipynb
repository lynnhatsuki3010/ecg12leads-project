```python
import os, sys
sys.path.append("../")

import numpy as np
import torch
from torch.utils.data import DataLoader

# --- Import các module quan trọng ---
from training.dataset_loader import ECGDataset
from training.balanced_sampler import create_balanced_dataloader

# Import hàm train mới (đã có Mixup & SWA)
from training.train_improved import train_model 

# Import mô hình mới (ResNet18 + LSTM)
from training.model_resnet18_lstm import ResNet18_LSTM_Attn, ResNet18_Pure, ResNet18_Attention 

# Import hàm evaluate mới (Sliding Window)
from training.evaluate import evaluate_model_sliding_window

#Import visualize
from visualize.visualizaton import (
    plot_confusion_matrix, 
    plot_training_history, 
    plot_roc_curves, 
    plot_per_class_metrics
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"✅ Thiết bị sử dụng: {device}")
```

### **5 LABELS**

```python
target_names = [
    "SB",
    "SR",
    "AF",
    "SVT",
    "ST",
]
```

```python
# === Đọc dữ liệu từ folder mới (Đã Fix length + Z-score) ===
# Lưu ý: Sửa đường dẫn trỏ về folder splits_fixed_v3
data_dir = r"E:\NCKH - 2026\ECG Project\data\processed\splits_leads12_160125"

print(f"Đang đọc dữ liệu từ: {data_dir}")

X_train = np.load(os.path.join(data_dir, "train_files.npy"), allow_pickle=True)
X_val = np.load(os.path.join(data_dir, "val_files.npy"), allow_pickle=True)
X_test = np.load(os.path.join(data_dir, "test_files.npy"), allow_pickle=True)

y_train = np.load(os.path.join(data_dir, "y_train.npy"))
y_val = np.load(os.path.join(data_dir, "y_val.npy"))
y_test = np.load(os.path.join(data_dir, "y_test.npy"))

# Load tên các nhãn để in báo cáo cho đẹp
try:
    target_names = np.load(os.path.join(data_dir, "target_names.npy"), allow_pickle=True).tolist()
except:
    # Fallback nếu không tìm thấy file tên
    target_names = [f"Class {i}" for i in range(y_train.shape[1])]

print(f" Thống kê dữ liệu:")
print(f"   Train: {len(X_train)} samples")
print(f"   Val  : {len(X_val)} samples")
print(f"   Test : {len(X_test)} samples")
print(f"   Số lớp (Classes): {y_train.shape[1]}")
print(f"   Tên lớp: {target_names}")
```

### **RESNET 18_PURE**

```python
# === Cấu hình Hyperparameters ===
BATCH_SIZE = 64        # Tăng lên 64 để train nhanh hơn và ổn định Batch Norm
LEARNING_RATE = 1e-3   
NUM_EPOCHS = 50        

# === Tạo Dataset & DataLoader ===
# Augment=True cho tập train để tăng tính đa dạng
train_dataset = ECGDataset(X_train, y_train, augment=True)
val_dataset = ECGDataset(X_val, y_val, augment=False) # Val không augment
test_dataset = ECGDataset(X_test, y_test, augment=False)

# DataLoader cho Train: Dùng Balanced Sampler
print(" Đang tạo Balanced DataLoader...")
train_loader = create_balanced_dataloader(
    train_dataset, y_train, 
    batch_size=BATCH_SIZE, 
    use_sampler=True, 
    oversample_factor=3.0  # Giảm xuống 3.0 (nhẹ nhàng hơn vì đã có Mixup)
)

# DataLoader cho Val/Test
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

print(" DataLoader đã sẵn sàng!")

# === Khởi tạo Model Mới ===
num_classes = y_train.shape[1]

# Gọi model ResNet18 + LSTM + Attention
model = ResNet18_Pure(num_classes=num_classes, input_channels=12)

print("\n Kiến trúc Model:")
# Đếm số lượng tham số để kiểm soát độ phức tạp
total_params = sum(p.numel() for p in model.parameters())
print(f"   Model: ResNet18_Pure")
print(f"   Tổng tham số (Parameters): {total_params:,}")
```

```python
# === Bắt đầu Huấn luyện ===
print("🚀 Bắt đầu quá trình huấn luyện...")
print(f"   Device: {device}")
print(f"   Strategy: Class-Balanced Focal Loss + Mixup + SWA")

model, history = train_model(
    model, 
    train_loader, 
    val_loader, 
    y_train,
    num_epochs=NUM_EPOCHS,
    patience=10,           # Kiên nhẫn đợi 10 epochs nếu loss không giảm
    lr=LEARNING_RATE,
    device=device,
    
    # --- Cấu hình nâng cao ---
    loss_type="cb_focal",     # Class-Balanced Focal Loss
    focal_gamma=2.0,
    
    use_mixup=True,           # BẬT Mixup
    mixup_alpha=0.4,          # Alpha 0.4 là chuẩn cho dataset y tế
    
    use_swa=True,             # BẬT SWA (Stochastic Weight Averaging)
    swa_start_epoch=30,        # Bắt đầu SWA từ epoch 30 (khi model đã hội tụ cơ bản)
    target_names=target_names
)

# === Lưu Model ===
model_save_path = r"E:\NCKH - 2026\ECG Project\models\resnet18_pure_260226.pth"
os.makedirs(os.path.dirname(model_save_path), exist_ok=True)
torch.save(model.state_dict(), model_save_path)
print(f"\n Đã lưu model tốt nhất tại: {model_save_path}")
```

```python
from training.evaluate import evaluate_model, evaluate_model_sliding_window

# Cách 1: Dùng hàm chuẩn (nhanh, giống cũ)
preds_bin, probs, cms = evaluate_model(
    model, test_loader, y_test, 
    device=device, 
    target_names=target_names
)
```

```python
# === Đánh giá mô hình trên tập Test ===
print("\nĐánh giá kết quả trên tập TEST (Sử dụng Sliding Window / Full Signal)...")

# Load lại model tốt nhất (nếu cần thiết, dù biến 'model' hiện tại đã là best)
# model.load_state_dict(torch.load(model_save_path))

# Gọi hàm đánh giá mới
preds_bin, preds_prob = evaluate_model_sliding_window(
    model, 
    test_loader, 
    y_test, 
    device=device, 
    target_names=target_names
)
```

```python
# Cell 1 - CẬP NHẬT
import os, sys
sys.path.append("../")

import numpy as np
import torch
from torch.utils.data import DataLoader

# Import modules
from training.dataset_loader import ECGDataset
from training.balanced_sampler import create_balanced_dataloader
from training.train_improved import train_model  # ← Giữ nguyên

# ========== THÊM IMPORTS MỚI ==========
from training.hybrid_dataset import HybridECGDataset  # Dataset mới
from training.model_hybrid import HybridECGModel      # Model mới

# Import model cũ (để so sánh hoặc fallback)
from training.model_resnet18_lstm import ResNet18_LSTM_Attn

# Import hàm evaluate
from training.evaluate import evaluate_model_sliding_window

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Thiết bị sử dụng: {device}")
```

```python
target_namess = [
    "Nhịp chậm xoang -SB",
    "Nhịp xoang bình thường - SR",
    "Cuồng nhĩ - AF",
    "Rung nhĩ - AFIB",
    "Nhịp nhanh trên thất - SVT",
    "Nhịp xoang nhanh - ST",
    "Phì đại thất trái - LVH"
]
```

```python
# === Đọc dữ liệu từ folder mới (Đã Fix length + Z-score) ===
# Lưu ý: Sửa đường dẫn trỏ về folder splits_fixed_v3
data_dir = r"E:\NCKH - 2026\ECG Project\data\processed\splits_leads12_040125"

print(f"Đang đọc dữ liệu từ: {data_dir}")

X_train = np.load(os.path.join(data_dir, "train_files.npy"), allow_pickle=True)
X_val = np.load(os.path.join(data_dir, "val_files.npy"), allow_pickle=True)
X_test = np.load(os.path.join(data_dir, "test_files.npy"), allow_pickle=True)

y_train = np.load(os.path.join(data_dir, "y_train.npy"))
y_val = np.load(os.path.join(data_dir, "y_val.npy"))
y_test = np.load(os.path.join(data_dir, "y_test.npy"))

# Load tên các nhãn để in báo cáo cho đẹp
try:
    target_names = np.load(os.path.join(data_dir, "target_names.npy"), allow_pickle=True).tolist()
except:
    # Fallback nếu không tìm thấy file tên
    target_names = [f"Class {i}" for i in range(y_train.shape[1])]

print(f" Thống kê dữ liệu:")
print(f"   Train: {len(X_train)} samples")
print(f"   Val  : {len(X_val)} samples")
print(f"   Test : {len(X_test)} samples")
print(f"   Số lớp (Classes): {y_train.shape[1]}")
print(f"   Tên lớp: {target_names}")
```

```python
# Cell 4 - HYBRID MODEL VERSION
# === Cấu hình ===
BATCH_SIZE = 16
LEARNING_RATE = 1e-3
NUM_EPOCHS = 50
USE_HYBRID = True  # ← SWITCH: True = Hybrid, False = Standard

# === Load paths cho hand features ===
if USE_HYBRID:
    signal_dir = r"E:\NCKH - 2026\ECG Project\data\processed\ecg_leads12_040125"
    hand_feat_dir = r"E:\NCKH - 2026\ECG Project\data\processed\hand_features_only_100125"
    
    # Tạo paths cho hand features
    train_hand_files = [f.replace(signal_dir, hand_feat_dir) for f in X_train]
    val_hand_files = [f.replace(signal_dir, hand_feat_dir) for f in X_val]
    test_hand_files = [f.replace(signal_dir, hand_feat_dir) for f in X_test]
    
    # Create Hybrid Datasets
    train_dataset = HybridECGDataset(
        signal_paths=X_train,
        hand_feature_paths=train_hand_files,
        labels=y_train,
        augment=True
    )
    
    val_dataset = HybridECGDataset(
        signal_paths=X_val,
        hand_feature_paths=val_hand_files,
        labels=y_val,
        augment=False
    )
    
    test_dataset = HybridECGDataset(
        signal_paths=X_test,
        hand_feature_paths=test_hand_files,
        labels=y_test,
        augment=False
    )
    
    # Create Hybrid Model
    num_classes = y_train.shape[1]
    model = HybridECGModel(
        num_classes=num_classes,
        input_channels=12,
        num_hand_features=64,
        fusion_dim=256
    )
    
    print("Using HYBRID Model (CNN + Hand-crafted Features)")

else:
    # Standard model (original code)
    train_dataset = ECGDataset(X_train, y_train, augment=True)
    val_dataset = ECGDataset(X_val, y_val, augment=False)
    test_dataset = ECGDataset(X_test, y_test, augment=False)
    
    num_classes = y_train.shape[1]
    model = ResNet18_LSTM_Attn(num_classes=num_classes, input_channels=12)
    
    print("Using STANDARD Model (ResNet-LSTM-Attn)")

# === DataLoaders (GIỮ NGUYÊN logic) ===
train_loader = create_balanced_dataloader(
    train_dataset, y_train, 
    batch_size=BATCH_SIZE, 
    use_sampler=True, 
    oversample_factor=3.0
)

val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

print(" DataLoader đã sẵn sàng!")
print(f"\n Kiến trúc Model:")
total_params = sum(p.numel() for p in model.parameters())
print(f"   Model: {type(model).__name__}")
print(f"   Tổng tham số: {total_params:,}")
```

```python

# ========== GIẢI PHÁP 1: KIỂM TRA & FIX SỐ FEATURES ==========

# Thêm cell này TRƯỚC khi khởi tạo model
import numpy as np

# Load 1 sample hand feature để kiểm tra
sample_hand_feat = np.load(train_hand_files[0])
ACTUAL_NUM_FEATURES = sample_hand_feat.shape[0]

print(f"✅ Số features thực tế: {ACTUAL_NUM_FEATURES}")
print(f"   Shape của hand features: {sample_hand_feat.shape}")
```

```python
# Cell 5 - TRAINING (CHỈ THÊM 1 DÒNG)
print(f"   Device: {device}")
print(f"   Strategy: Class-Balanced Focal Loss + Mixup + SWA")

model = train_model(
    model, 
    train_loader, 
    val_loader, 
    y_train,
    num_epochs=NUM_EPOCHS,
    patience=10,
    lr=LEARNING_RATE,
    device=device,
    
    # Cấu hình nâng cao (GIỮ NGUYÊN)
    loss_type="cb_focal",
    focal_gamma=2.0,
    use_mixup=True,
    mixup_alpha=0.4,
    use_swa=True,
    swa_start_epoch=30,
    
    # ========== THÊM DÒNG NÀY ==========
    is_hybrid=USE_HYBRID  # ← True nếu dùng HybridECGModel
)

# Lưu model
model_save_path = r"E:\NCKH - 2026\ECG Project\models\hybrid_model_090125.pth"
os.makedirs(os.path.dirname(model_save_path), exist_ok=True)
# Check if SWA wrapped model
if hasattr(model, 'module'):
    # It's an AveragedModel (SWA) - save the underlying module
    torch.save(model.module.state_dict(), model_save_path)
    print(f"\nSaved SWA model to: {model_save_path}")
else:
    # Standard model
    torch.save(model.state_dict(), model_save_path)
    print(f"\nSaved model to: {model_save_path}")
```

```python
# === Đọc dữ liệu từ folder mới (Đã Fix length + Z-score) ===
# Lưu ý: Sửa đường dẫn trỏ về folder splits_fixed_v3
data_dir = r"E:\NCKH - 2026\ECG Project\data\processed\splits_leads12_040125"

print(f"Đang đọc dữ liệu từ: {data_dir}")

X_train = np.load(os.path.join(data_dir, "train_files.npy"), allow_pickle=True)
X_val = np.load(os.path.join(data_dir, "val_files.npy"), allow_pickle=True)
X_test = np.load(os.path.join(data_dir, "test_files.npy"), allow_pickle=True)

y_train = np.load(os.path.join(data_dir, "y_train.npy"))
y_val = np.load(os.path.join(data_dir, "y_val.npy"))
y_test = np.load(os.path.join(data_dir, "y_test.npy"))

# Load tên các nhãn để in báo cáo cho đẹp
try:
    target_names = np.load(os.path.join(data_dir, "target_names.npy"), allow_pickle=True).tolist()
except:
    # Fallback nếu không tìm thấy file tên
    target_names = [f"Class {i}" for i in range(y_train.shape[1])]

print(f" Thống kê dữ liệu:")
print(f"   Train: {len(X_train)} samples")
print(f"   Val  : {len(X_val)} samples")
print(f"   Test : {len(X_test)} samples")
print(f"   Số lớp (Classes): {y_train.shape[1]}")
print(f"   Tên lớp: {target_names}")
```

```python
# === Cấu hình Hyperparameters ===
BATCH_SIZE = 16        # Tăng lên 64 để train nhanh hơn và ổn định Batch Norm
LEARNING_RATE = 1e-3   
NUM_EPOCHS = 50        

# === Tạo Dataset & DataLoader ===
# Augment=True cho tập train để tăng tính đa dạng
train_dataset = ECGDataset(X_train, y_train, augment=True)
val_dataset = ECGDataset(X_val, y_val, augment=False) # Val không augment
test_dataset = ECGDataset(X_test, y_test, augment=False)

# DataLoader cho Train: Dùng Balanced Sampler
print(" Đang tạo Balanced DataLoader...")
train_loader = create_balanced_dataloader(
    train_dataset, y_train, 
    batch_size=BATCH_SIZE, 
    use_sampler=True, 
    oversample_factor=3.0  # Giảm xuống 3.0 (nhẹ nhàng hơn vì đã có Mixup)
)

# DataLoader cho Val/Test
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

print(" DataLoader đã sẵn sàng!")

# === Khởi tạo Model Mới ===
num_classes = y_train.shape[1]

# Gọi model ResNet18 + LSTM + Attention
model = ResNet18_LSTM_Attn(num_classes=num_classes, input_channels=12)

print("\n Kiến trúc Model:")
# Đếm số lượng tham số để kiểm soát độ phức tạp
total_params = sum(p.numel() for p in model.parameters())
print(f"   Model: ResNet18_LSTM_Attn")
print(f"   Tổng tham số (Parameters): {total_params:,}")
```

```python
# === Bắt đầu Huấn luyện ===
print("🚀 Bắt đầu quá trình huấn luyện...")
print(f"   Device: {device}")
print(f"   Strategy: Class-Balanced Focal Loss + Mixup + SWA")

model = train_model(
    model, 
    train_loader, 
    val_loader, 
    y_train,
    num_epochs=NUM_EPOCHS,
    patience=10,           # Kiên nhẫn đợi 10 epochs nếu loss không giảm
    lr=LEARNING_RATE,
    device=device,
    
    # --- Cấu hình nâng cao ---
    loss_type="cb_focal",     # Class-Balanced Focal Loss
    focal_gamma=2.0,
    
    use_mixup=True,           # BẬT Mixup
    mixup_alpha=0.4,          # Alpha 0.4 là chuẩn cho dataset y tế
    
    use_swa=True,             # BẬT SWA (Stochastic Weight Averaging)
    swa_start_epoch=30        # Bắt đầu SWA từ epoch 30 (khi model đã hội tụ cơ bản)
)

# === Lưu Model ===
model_save_path = r"E:\NCKH - 2026\ECG Project\models\resnet18_lstm_mixup_swa_060125_BATCH16.pth"
os.makedirs(os.path.dirname(model_save_path), exist_ok=True)
torch.save(model.state_dict(), model_save_path)
print(f"\n Đã lưu model tốt nhất tại: {model_save_path}")
```

```python
def load_model_safe(model, model_path, device="cuda"):
    """
    Load model state dict an toàn, tự động xử lý SWA models.
    
    Args:
        model: Model instance (standard)
        model_path: Path to .pth file
        device: Device to load to
        
    Returns:
        model: Loaded model
    """
    # Load state dict
    state_dict = torch.load(
        model_path, 
        map_location=device,
        weights_only=False
    )
    
    # Check if SWA model
    is_swa = 'n_averaged' in state_dict
    
    if is_swa:
        print("⚠️  Detected SWA model")
        # Remove SWA metadata
        state_dict.pop('n_averaged', None)
        
        # Remove 'module.' prefix
        new_state_dict = {}
        for key, value in state_dict.items():
            new_key = key.replace('module.', '') if key.startswith('module.') else key
            new_state_dict[new_key] = value
        
        state_dict = new_state_dict
        print("✅ Cleaned SWA state dict")
    
    # Load
    model.load_state_dict(state_dict)
    model.to(device)
    model.eval()
    
    print(f"✅ Model loaded from: {model_path}")
    return model


# Usage:
model_path = r"E:\NCKH - 2026\ECG Project\models\resnet18_lstm_mixup_swa_040125.pth"
model = load_model_safe(model, model_path, device)
```

```python
#threshold
from training.threshold_tuning import adaptive_threshold_search

best_thresholds = adaptive_threshold_search(
    model, 
    val_loader, 
    device=device, 
    num_classes=y_train.shape[1],
    verbose=True  # In ra bảng chi tiết để bạn thấy ngưỡng thay đổi thế nào
)

print(f"\nNgưỡng tối ưu tìm được: {best_thresholds}")
```

```python
# Gọi hàm đánh giá mới
preds_bin, preds_prob = evaluate_model_sliding_window(
    model, 
    test_loader, 
    y_test, 
    device=device,
    best_thresholds=best_thresholds,
    target_names=target_names
)
```
